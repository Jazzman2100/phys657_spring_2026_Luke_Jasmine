```python
# Module 2 Activity 5

# By Jasmine

import numpy as np
import matplotlib.pyplot as plt
size = 100
mu = [0, 0]
cov = [[4, 2], [2,16]]
data = np.random.multivariate_normal(mu, cov, size)
x, y = data.T
plt.plot(x, y, 'x')
plt.show()
```

```python
# Sequential Estimation

# Starting with guesses of mu and covariance (let both be zero)
mu_est = np.zeros(2) 
cov_est = np.zeros((2, 2))

mu_history = []

for n in range(1, size + 1):
    x_n = data[n-1] # Get the current observation
    mu_est = mu_est + (1/n) * (x_n - mu_est)
    
    if n > 1:
        diff = x_n - mu_est
        cov_est = ((n-1)/n) * cov_est + (1/n) * np.outer(diff, diff)
    
    mu_history.append(mu_est.copy())

print(f"True Mean: {mu}")
print(f"Estimated Mean: {mu_est}")
print(f"\nTrue Covariance:\n{cov}")
print(f"Estimated Covariance:\n{cov_est}")
```

Since our covariance matrix is based on estimated $\mu$, and the $\sigma_{x} + \sigma_{y}$ are elements of the covariance matrix, and we have biases on estimating this covariance matrix with less data points. 
<br> **Reasoning**: less data points give less accurate mean, and therefore it impacts the accuracy of estimating the covariance matrix. 
<br> **How to fix it** : get more data points. According to $Eq(2.125)$, we can see that as $N$ goes larger, $\frac{1}{N-1}$ gets smaller, which means the bias gets smaller.
