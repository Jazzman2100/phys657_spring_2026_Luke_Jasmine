# Activity 4: Maximum Likelihood Estimation (by hand) for a Gaussian

Assume a dataset $\{x_n\}_{n=1}^N$ is drawn independently from a Gaussian distribution with unknown mean $\mu$ and variance $\sigma^2$.

1. Write down the likelihood function $p(\{x_n\} \mid \mu, \sigma^2)$.
2. Compute the log-likelihood.
3. Derive the maximum likelihood estimator for $\mu$.
4. Derive the maximum likelihood estimator for $\sigma^2$.
5. Compare your expression for $\sigma^2$ with the unbiased sample variance estimator. Briefly explain the difference, and discuss when this difference is relevant.

### (1) (2) : Gaussian Distribution
$$\mathcal{N}(x|\mu, \sigma^2) = \frac{1}{(2\pi\sigma^2)^{1/2}} \exp \left\{ -\frac{1}{2\sigma^2}(x-\mu)^2 \right\}$$
$\mu, \sigma \text{ unknown for }
\{x_n\}_{n=1}^N \text{ data set}$

---

$$\text{The likelihood function can be written as: }p(\{x_n\} | \mu, \sigma^2) = \prod_{n=1}^N \mathcal{N}(x_n | \mu, \sigma^2)$$

$$\text{We take the natural log of the likelihood function and use the rule of log to get: }$$

$$\sum_{i=1}^N \ln \left( \frac{1}{\sqrt{2\pi\sigma^2}} e^{\left\{ -\frac{1}{2\sigma^2}(x_i-\mu)^2 \right\}} \right)$$
$$\Rightarrow \sum \left( \ln \frac{1}{\sqrt{2\pi\sigma^2}} + \ln e^{\left\{ -\frac{1}{2\sigma^2}(x-\mu)^2 \right\}} \right)$$
$$\Rightarrow \sum \left( -\frac{1}{2} \ln 2\pi\sigma^2 + -\frac{1}{2\sigma^2}(x-\mu)^2 \right)$$
$$= \underbrace{\sum -\frac{1}{2} \ln(2\pi)} - \frac{1}{2} \ln \sigma^2 - \frac{N}{2\sigma^2}(x-\mu)^2$$

$$\text{Hence, the log likelikhood is: }ln(p) = -\frac{N}{2} \ln(2\pi) - \frac{N}{2} \ln \sigma^2 - \frac{1}{2\sigma^2} \sum_{i=1}^N (x_n - \mu)^2$$

---

### (3) (4) : Maximum in terms of $\mu, \sigma^2$

The maximum likelihood estimator for $\mu$: 

$$\frac{dp}{d\mu} = -\frac{1}{2\sigma^2} \sum_{i=1}^N 2(x_n - \mu)(-1)$$

Let $\frac{dp}{d\mu} = 0 \Rightarrow +\frac{2}{2\sigma^2} \sum_{i=1}^N (x_n - \mu) = 0$
$\Rightarrow \underbrace{\sum_{i=1}^N (x_n - \mu) = 0}$
$\Rightarrow \sum_{i=1}^N x_n - N\mu = 0$

$$\frac{1}{N} \sum_{i=1}^N x_n = \mu$$ -- *Eq(1.55)*

---
The maximum likelihood estimator for $\sigma^2$: 

$$\frac{dp}{d\sigma^2} = -\frac{N}{2} \frac{1}{\sigma^2} - (-1) \frac{1}{2} \frac{1}{(\sigma^2)^2} \sum_{i=1}^N (x_n - \mu)^2$$

To find maximum $\Rightarrow$ we let $\frac{dp}{d\sigma^2} = 0$

$$\frac{N}{\sigma^2} = \left( \frac{1}{2} \sum (x_n - \mu)^2 \right) \frac{1}{(\sigma^2)^2} \times 2$$
$$\Rightarrow \underbrace{\sigma^2 = \frac{1}{N} \sum_{i=1}^N (x_n - \mu)^2} \text{, compare with}$$

$$(eq 1.59): \tilde{\sigma}^2 = \frac{1}{N-1} \sum_{i=1}^N (x_n - \mu)^2$$

${\sigma^2}$ is the sample variance measured with respect to sample mean $\mu$.

We discuss the difference between $\sigma^2$ & $\tilde{\sigma}^2$: 

$$\underbrace {\sigma_{ML}^2} = \left( \frac{N-1}{N} \right) \tilde{\sigma}^2$$
$$\sigma_{ML}^2 = \frac{1}{N} \sum_{n=1}^{N} (x_n - \mu)^2 = \left( \frac{N-1}{N} \right) \tilde{\sigma}^2$$

$$\Rightarrow \tilde{\sigma}^2 = \frac{1}{N-1} \sum_{i=1}^N (x_n - \mu)^2$$
