```python
# Module 2 Activity 3

# Problem 1
# By Jasmine

import numpy as np
import matplotlib.pyplot as plt

# List out the known parameters
mean = [0, 0]
cov = [[4, 2], [2, 16]]

# From this we know sigma_x^2 = 4 and sigma_y^2 = 16, so sigma_x = 2 and sigma_y = 4. 
# Since the (1,2) element and the (2,1) element are both 2 and they both represent rho*sigma_x*sigma_y, we know rho = 1/4. 

n_samples = 1000

# Generate the data
data = np.random.multivariate_normal(mean, cov, n_samples)

x, y = data.T

# Plots
plt.figure(figsize=(8, 6))
plt.scatter(x, y, alpha=1, s=8)
plt.title(f"Generated 2D Gaussian PDF($\sigma_x=2, \sigma_y=4, \\rho=0.25$)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# Problem 2

from scipy.stats import multivariate_normal

# Create grid for contour
x_graph = np.linspace(-15, 15, 100)
y_graph = np.linspace(-15, 15, 100)
X_graph, Y_graph = np.meshgrid(x_graph, y_graph)

# Calculate Z values (Probability Density)
pos = np.dstack((X_graph, Y_graph))
rv = multivariate_normal(mean, cov)
z_graph = rv.pdf(pos)

# Plots
fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(x, y, s=2, alpha=1, label="Data")

contour = ax.contour(X_graph, Y_graph, z_graph, levels=8, linewidths=1.5)

ax.set_xlim(-15, 15)
ax.set_ylim(-15, 15)
ax.set_title("2D Gaussian data plot with contours")
ax.set_xlabel("Variable X")
ax.set_ylabel("Variable Y")
plt.legend()
plt.show()
```
