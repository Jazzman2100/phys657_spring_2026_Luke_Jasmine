```python
# Module 2 Activity 6

# Problem 1
# By Jasmine

# Gaussian has mean (mu), sigma (std), and the number of samples.

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu = 0          
sigma = 0.8
N = 20

samples = np.random.normal(mu, sigma, N)

print(f"Generated {N} samples from Gaussian(μ={mu}, σ={sigma}):\n")
print(samples)
```

```python
# Trial values are mu: we want to see how the log-likelihood function plots with trial mu values.

mu_grid = np.linspace(-5, 5, 1000)

# Problem 2: compute the log-likelihood function
def compute_log_likelihood(mu_trial, data, sigma):
    n = len(data)
    term1 = -n/2 * np.log(2 * np.pi)
    term2 = -n/2 * np.log(sigma**2)
    term3 = -1/(2 * sigma**2) * np.sum((data - mu_trial)**2)
    return term1 + term2 + term3
    
log_likelihoods = [compute_log_likelihood(m, samples, sigma) for m in mu_grid]

# Find the Maximum Likelihood Estimate (MLE) from our grid
maxlikelihood_mu = mu_grid[np.argmax(log_likelihoods)]
print(maxlikelihood_mu)

# Problem 3: plot the log-likelihood vs trial mu val
# Plots
plt.figure(figsize=(10, 5))
plt.plot(mu_grid, log_likelihoods, color='green')
plt.axvline(maxlikelihood_mu, color='black', linestyle='--', label='Trial $\mu$ value that maximize the log-likelihood function')
plt.title('Log-Likelihood Function vs. $\mu$')
plt.xlabel('Trial $\mu$ values')
plt.ylabel('Log-Likelihood')
plt.legend()
plt.show()
```
