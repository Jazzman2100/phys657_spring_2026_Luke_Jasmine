```python
# Module 2 Activity 1

# By Jasmine

import numpy as np
import matplotlib.pyplot as plt

N = 1     

# trials are how many times that we compute the average
trials = 1000  

x_list = []
for i in range (N): 
    xi = np.random.uniform(0,1)
    i = i + 1
    x_list.append(xi)

x_avg_list = []

# distribution of averages
for _ in range(trials):
    samples = np.random.uniform(0, 1, N)
    current_avg = np.mean(samples)
    x_avg_list.append(current_avg)

# Plots
plt.hist(x_avg_list, bins=10, density=True, color='skyblue', edgecolor='black')
plt.title(f'Distribution of x_avg for N = {N}')
plt.xlabel('Value of x_avg')
plt.ylabel('Frequency')

# Gaussian fit attempt with hard-coded, guessed parameters
mu, sigma = 0.5, np.sqrt((1/12) / N)
x = np.linspace(0, 1, 100)
p = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma)**2)
# Luke uses scipy
plt.plot(x, p, 'r', linewidth=2, label='Gaussian PDF')
plt.legend()

plt.show()
```
